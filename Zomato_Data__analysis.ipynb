```python
import pandas as pd
import numpy as np

# Load the data
df = pd.read_csv('zomato.csv')  # Replace with your actual file name

# Inspect the 'rate' column
print(df['rate'].unique())

# Clean the 'rate' column
# First, replace non-numeric entries with NaN
df['rate'] = df['rate'].where(df['rate'].str.match(r'^\d+(\.\d+)?/\d+$'), other=np.nan)

# Drop rows with NaN values in the 'rate' column
df.dropna(subset=['rate'], inplace=True)

# Now split and convert the rate column to float
df['rate'] = df['rate'].str.split('/').str[0].astype(float)

# Calculate average rating by cuisine
avg_rating = df.groupby('cuisines')['rate'].mean().sort_values(ascending=False)

# Print the average rating
print(avg_rating)

# Plotting (if needed)
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 6))
sns.barplot(x=avg_rating.index, y=avg_rating.values)
plt.xticks(rotation=45)
plt.title('Average Rating by Cuisine')
plt.xlabel('Cuisine')
plt.ylabel('Average Rating')
plt.show()
```

```python
# Distribution of Ratings
plt.figure(figsize=(10, 6))
sns.histplot(df['rate'], bins=20, kde=True)
plt.title('Distribution of Ratings')
plt.xlabel('Rating')
plt.ylabel('Frequency')
plt.axvline(df['rate'].mean(), color='r', linestyle='--', label='Mean Rating')
plt.legend()
plt.show()
```

```python
# Clean and convert the approx_cost column
df['approx_cost(for two people)'] = df['approx_cost(for two people)'].replace({'₹': '', ',': ''}, regex=True).astype(float)

# Calculate average cost by cuisine
avg_cost = df.groupby('cuisines')['approx_cost(for two people)'].mean().sort_values(ascending=False)

# Plot average cost by cuisine
plt.figure(figsize=(12, 6))
sns.barplot(x=avg_cost.index, y=avg_cost.values)
plt.xticks(rotation=45)
plt.title('Average Cost by Cuisine')
plt.xlabel('Cuisine')
plt.ylabel('Average Cost (for two people)')
plt.show()
```
